```python
import math as mt
import numpy as np
import matplotlib.pylab as pl
```

```python
def osci(x):
    w0 = 1
    y = np.zeros(2)
    y[0] = x[1]
    y[1] = -w0*x[0]
    return y
```

```python
x = np.linspace(-10,10,10)
y = np.linspace(-10,10,10)

X = []
for i in x:
    for j in y:
        x = np.array([i,j])
        u = osci(x)
        pl.quiver(x[0],x[1],u[0],u[1])
pl.axhline(y=0,c='k')
pl.axvline(x=0,c='k')
pl.grid()
```

```python
def Euler2D(dx,r0,t0,t1,h): # En este problema en particular consideramos el valor RC
  T = [t0]
  X = [r0[0]]
  Y = [r0[1]]
  while t0 < t1:
    r = r0 + dx(r0)*h
    t = t0 + h
    T.append(t)
    X.append(r[0])
    Y.append(r[1])
    r0 = r
    t0 = t
  return T,X,Y
```

```python
t0 = 1
t1 = 100
h = 0.01

t,pos,vel = Euler2D(osci,[0,1],t0,t1,h)
pl.plot(t,pos)
pl.plot(t,vel)
```

Notemos que hay algo creciente, esto se debe al método de Euler, el de Runge-Kutta funciona mejor.

Ahora con el de Runge-Kutta:

```python
def RungeKutta2D(dx,r0,t0,t1,h):
  T = [t0]
  X = [r0[0]]
  Y = [r0[1]]
  while t0 < t1:
    k1 = dx(r0)*h
    k2 = dx(r0+0.5*k1)*h # A diferencia del ejercicio anterior, la ecuación que
    k3 = dx(r0+0.5*k2)*h # queremos resolve no depende de t, así que en este caso
    k4 = dx(r0+k3)*h # omitimos el incremento en esa variable

    r = r0 + (k1 + 2*k2 + 2*k3 + k4)/6
    t = t0 + h
    T.append(t)
    X.append(r[0])
    Y.append(r[1])
    r0 = r
    t0 = t
  return T,X,Y
```

```python
t0 = 1
t1 = 100
h = 0.01

t,pos,vel = RungeKutta2D(osci,[0,1],t0,t1,h)
pl.plot(t,pos)
pl.plot(t,vel)
```
