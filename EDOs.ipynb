### Método de Euler

Dada una condición inicial $x(t)$ para alguna $t$, podemos escribir el valor de $x$ en un intervalo $h$ usando una expansión de Taylor (alrededor de t):

$$\begin{align} x(t+h) &= x(t) + h\frac{dx}{dt} + \frac{1}{2}h^2\frac{d^2x}{dt^2} + ...
\\ &= x(t) + hf(x,t) + O(h^2)
\\ &\text{pero si $h^2$ es pequeña entonces podemos despreciar $O$, así que obtenemos}
\\ x(t+h) &= x(t) + hf(x,t) \end{align}$$

EJEMPLO

```python
import math as mt
import numpy as np
import matplotlib.pyplot as pl
```

```python
def f(x,t):
  return -x**3 + mt.sin(t)
```

```python
a = 0 # Inicio del intervalo
b = 10 # Fin del intervalo
N = 100 # Número de pasos
h = (b-a)/N # Tamaño de paso
x = 0 # Condición incial

T = np.arange(a,b,h)
X = []
for t in T:
  X.append(x) # Primero ya que tenemos la condición inicial
  x = x + h*f(x,t)

pl.plot(T,X)
pl.xlabel('t')
pl.ylabel('x(t)')
pl.show()
```
